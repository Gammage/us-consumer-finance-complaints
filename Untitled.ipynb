```python
import pandas as pd
import numpy
import matplotlib.pyplot as plt
import plotly.express as px
import sqlite3

conn = sqlite3.connect("./data/database.sqlite")
```

```python
pd.read_sql(" PRAGMA table_info(consumer_complaints) ", conn)
```

```python
pd.read_sql(" SELECT product, COUNT(complaint_id) FROM consumer_complaints GROUP BY product ORDER BY COUNT(complaint_id) DESC LIMIT 10", conn)
```

```python

# complaints by product 

comp_prod = pd.read_sql(" SELECT product, COUNT(complaint_id) AS count FROM consumer_complaints GROUP BY product ORDER BY count ASC LIMIT 10", conn)
ax = comp_prod.set_index('product').plot(kind='barh')
ax.set_facecolor("none")
ax.tick_params(colors="#c0f0e0")
for spine in ax.spines.values():
    spine.set_color("#33666b")
ax.title.set_color("#c0f0e0")
ax.xaxis.label.set_color("#c0f0e0")
ax.yaxis.label.set_color("#c0f0e0")
plt.xlabel('Number of complaints')
plt.savefig("output/product_complaints.png", bbox_inches="tight", facecolor="#00343d")
plt.show()
```

```python
# complaints by state

comp_state = pd.read_sql(" SELECT state, COUNT(complaint_id) AS count FROM consumer_complaints GROUP BY state ORDER BY count", conn)
fig = px.choropleth(comp_state, locations="state", locationmode='USA-states', color="count", template='plotly_dark', color_continuous_scale='blues').update_layout(geo_scope='usa')
fig.write_html("output/state_complaints.html")
```

```python
# timely company complaint response rate

comp_response = pd.read_sql(""" SELECT company, total, timely, (timely * 100.0 / total) AS pct
FROM (
    SELECT company,
        COUNT(*) AS total,
        SUM(CASE WHEN timely_response = 'Yes' THEN 1 ELSE 0 END) AS timely
    FROM consumer_complaints
    GROUP BY company
    HAVING total > 100
    )
    ORDER BY pct ASC
    LIMIT 10 """, conn)
ax = comp_response.set_index('company')['pct'].plot(kind='barh')
ax.set_facecolor("none")
ax.tick_params(colors="#c0f0e0")
for spine in ax.spines.values():
    spine.set_color("#33666b")
ax.title.set_color("#c0f0e0")
ax.xaxis.label.set_color("#c0f0e0")
ax.yaxis.label.set_color("#c0f0e0")
plt.xlabel("Timely response response rate %")
print(comp_response)
plt.savefig("output/company_response_rate.png", bbox_inches="tight", facecolor="#00343d")
plt.show()
```

```python
# monthly complaint trend

df = pd.read_sql("SELECT date_received, complaint_id FROM consumer_complaints", conn)
df['date_received'] = pd.to_datetime(df['date_received'])
df['month'] = df['date_received'].dt.to_period('M')
monthly = df.groupby('month')['complaint_id'].count()
ax = monthly.plot()
ax.set_facecolor("none")
ax.tick_params(colors="#c0f0e0")
for spine in ax.spines.values():
    spine.set_color("#33666b")
ax.title.set_color("#c0f0e0")
ax.xaxis.label.set_color("#c0f0e0")
ax.yaxis.label.set_color("#c0f0e0")
plt.xlabel('year')
plt.ylabel('Complaints')
plt.savefig("output/monthly_complaints.png", bbox_inches="tight", facecolor="#00343d")
plt.show()
```

```python
# dispute rate by submission channel

df = pd.read_sql(""" SELECT submitted_via, disputed, (disputed * 100.0 / total ) AS pct
                 FROM (
                 SELECT submitted_via,
                 COUNT(*) AS total,
                 SUM(CASE WHEN "consumer_disputed?" = 'Yes' THEN 1 ELSE 0 END) as disputed
                    FROM consumer_complaints 
                    GROUP BY submitted_via
                    )
                    ORDER BY pct ASC
                 """, conn)
ax = df.set_index('submitted_via')['pct'].plot(kind='barh')
ax.set_facecolor("none")
ax.tick_params(colors="#c0f0e0")
for spine in ax.spines.values():
    spine.set_color("#33666b")
ax.title.set_color("#c0f0e0")
ax.xaxis.label.set_color("#c0f0e0")
ax.yaxis.label.set_color("#c0f0e0")
plt.xlabel('Dispute rate %')
plt.savefig("output/dispute_rate_submission_channel.png", bbox_inches="tight", facecolor="#00343d")
plt.show()
```
